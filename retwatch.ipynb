```python
import csv
import json

def flatten_json(data, parent_key='', sep='.'):
    """
    Recursively flattens a nested dictionary or list.

    Parameters:
        data (dict or list): The JSON structure to flatten.
        parent_key (str): The prefix for keys (used during recursion).
        sep (str): Separator between keys.

    Returns:
        dict: A flattened dictionary where nested keys are concatenated.
    """
    items = {}
    if isinstance(data, dict):
        for key, value in data.items():
            new_key = f"{parent_key}{sep}{key}" if parent_key else key
            items.update(flatten_json(value, new_key, sep=sep))
    elif isinstance(data, list):
        for i, value in enumerate(data):
            new_key = f"{parent_key}{sep}{i}" if parent_key else str(i)
            items.update(flatten_json(value, new_key, sep=sep))
    else:
        items[parent_key] = data
    return items

def process_csv(input_file, output_file):
    """
    Processes an input CSV file cell-by-cell:
        - If a cell contains valid JSON (and that JSON is a dict or list),
          it flattens the JSON and prefixes the flattened keys with the original column name.
        - Otherwise, the cell's content is preserved.
    The function then writes all rows into a new CSV file with appropriate column headers.

    Parameters:
        input_file (str): Path to the input CSV file.
        output_file (str): Path to the output CSV file.
    """
    processed_rows = []    # List of dictionaries (one per row) after processing
    all_columns = set()    # To collect all column headers that will appear in the output

    # Open and read the input CSV file (assumed to have headers)
    with open(input_file, 'r', encoding='utf-8', newline='') as csv_in:
        reader = csv.DictReader(csv_in)
        for row in reader:
            new_row = {}
            # Process each cell in the row
            for col, cell in row.items():
                # If cell is nonempty, try to parse it as JSON
                if cell.strip():
                    try:
                        parsed = json.loads(cell)
                        # If the cell is valid JSON and it is a dict or list,
                        # flatten it.
                        if isinstance(parsed, (dict, list)):
                            flat = flatten_json(parsed)
                            # For each flattened key, prefix with the original column name.
                            for k, v in flat.items():
                                # For clarity, if k is empty, simply use the column name.
                                new_col = f"{col}{'.' if k else ''}{k}" if k else col
                                new_row[new_col] = v
                                all_columns.add(new_col)
                        else:
                            # For JSON values that are simple (e.g., numbers, strings), keep as is.
                            new_row[col] = parsed
                            all_columns.add(col)
                    except json.JSONDecodeError:
                        # If the cell is not valid JSON, simply keep the original value.
                        new_row[col] = cell
                        all_columns.add(col)
                else:
                    # If the cell is empty, simply pass it along.
                    new_row[col] = cell
                    all_columns.add(col)
            processed_rows.append(new_row)

    # Sort the column headers (this helps keep the output consistent)
    fieldnames = sorted(all_columns)

    # Write the flattened data to the output CSV file.
    with open(output_file, 'w', encoding='utf-8', newline='') as csv_out:
        writer = csv.DictWriter(csv_out, fieldnames=fieldnames)
        writer.writeheader()
        for row in processed_rows:
            writer.writerow(row)

if __name__ == '__main__':
    # Define your input and output CSV file names here.
    input_csv = 'scopus_retracted_articles.csv'
    output_csv = 'output.csv'

    process_csv(input_csv, output_csv)
    print(f"Processed CSV written to {output_csv}")
```

```python
import csv
import ast
import re

def flatten_json(nested_json, parent_key='', sep='.'):
    """
    Recursively flattens a nested JSON/dictionary.
    For lists, it uses the index as a key.
    """
    items = {}
    if isinstance(nested_json, dict):
        for k, v in nested_json.items():
            new_key = f"{parent_key}{sep}{k}" if parent_key else k
            items.update(flatten_json(v, new_key, sep=sep))
    elif isinstance(nested_json, list):
        for i, v in enumerate(nested_json):
            new_key = f"{parent_key}{sep}{i}" if parent_key else str(i)
            items.update(flatten_json(v, new_key, sep=sep))
    else:
        items[parent_key] = nested_json
    return items

def extract_link(link_list, ref="scopus"):
    """
    Given a list of link dictionaries, returns the @href value for the one whose @ref equals 'scopus'
    (or another value if specified).
    """
    if isinstance(link_list, list):
        for item in link_list:
            if isinstance(item, dict) and item.get('@ref') == ref:
                return item.get('@href', '')
    return ''

def process_entry(entry):
    """
    Processes a single paper record (an entry from the input data) by flattening it and extracting
    selected fields into a dictionary whose keys match the desired output columns.
    """
    # Flatten the entry
    flat = flatten_json(entry)

    out = {}
    # --- mapping keys to desired output columns ---
    out["Title"] = flat.get("dc:title", "")

    # Extract year from cover date (assumed format "YYYY-MM-DD")
    cover_date = flat.get("prism:coverDate", "")
    if cover_date:
        m = re.match(r"(\d{4})", cover_date)
        out["Year"] = m.group(1) if m else ""
    else:
        out["Year"] = ""

    out["Source title"] = flat.get("prism:publicationName", "")
    out["Volume"] = flat.get("prism:volume", "")
    out["Issue"] = flat.get("issue", "") or flat.get("prism:issueIdentifier", "")
    out["Art. No."] = flat.get("article-number", "")

    # Process page range
    page_range = flat.get("prism:pageRange", "")
    if page_range:
        if '-' in page_range:
            parts = page_range.split('-')
            out["Page start"] = parts[0].strip()
            out["Page end"] = parts[1].strip() if len(parts) > 1 else ""
            try:
                start = int(parts[0].strip())
                end = int(parts[1].strip())
                out["Page count"] = str(end - start + 1)
            except Exception:
                out["Page count"] = ""
        else:
            out["Page start"] = page_range
            out["Page end"] = ""
            out["Page count"] = ""
    else:
        out["Page start"] = ""
        out["Page end"] = ""
        out["Page count"] = ""

    out["Cited by"] = flat.get("citedby-count", "")
    out["DOI"] = flat.get("prism:doi", "")

    # Extract link (using the original entry if available)
    if "link" in entry:
        out["Link"] = extract_link(entry["link"], ref="scopus")
    else:
        out["Link"] = ""

    # Process affiliations
    if "affiliation" in entry:
        aff = entry["affiliation"]
        if isinstance(aff, list):
            aff_list = []
            for a in aff:
                if isinstance(a, dict):
                    parts = []
                    for key in ["affilname", "affiliation-city", "affiliation-country"]:
                        if key in a:
                            parts.append(a[key])
                    aff_list.append(", ".join(parts))
                else:
                    aff_list.append(str(a))
            out["Affiliations"] = "; ".join(aff_list)
        else:
            out["Affiliations"] = str(aff)
    else:
        out["Affiliations"] = ""

    out["PubMed ID"] = flat.get("pubmed-id", "")
    out["EID"] = flat.get("eid", "")
    out["Authors"] = flat.get("dc:creator", "")

    # For fields not available from the input, output empty strings.
    extra_fields = [
        "Author full names", "Author(s) ID", "Authors with affiliations", "Abstract",
        "Author Keywords", "Index Keywords", "Molecular Sequence Numbers", "Chemicals/CAS",
        "Tradenames", "Manufacturers", "Funding Details", "Funding Texts", "References",
        "Correspondence Address", "Editors", "Publisher", "Sponsors", "Conference name",
        "Conference date", "Conference location", "Conference code", "ISSN", "ISBN",
        "CODEN", "Language of Original Document", "Abbreviated Source Title",
        "Document Type", "Publication Stage", "Open Access", "Source"
    ]
    for field in extra_fields:
        out[field] = ""

    return out

def process_csv(input_file, output_file):
    """
    Reads an input CSV file where each row contains a Python dictionary literal (as a string),
    processes each record, and writes an output CSV file with the desired columns.
    """
    headers = [
        "Authors", "Author full names", "Author(s) ID", "Title", "Year",
        "Source title", "Volume", "Issue", "Art. No.", "Page start", "Page end",
        "Page count", "Cited by", "DOI", "Link", "Affiliations",
        "Authors with affiliations", "Abstract", "Author Keywords", "Index Keywords",
        "Molecular Sequence Numbers", "Chemicals/CAS", "Tradenames", "Manufacturers",
        "Funding Details", "Funding Texts", "References", "Correspondence Address",
        "Editors", "Publisher", "Sponsors", "Conference name", "Conference date",
        "Conference location", "Conference code", "ISSN", "ISBN", "CODEN",
        "PubMed ID", "Language of Original Document", "Abbreviated Source Title",
        "Document Type", "Publication Stage", "Open Access", "Source", "EID"
    ]

    output_rows = []

    with open(input_file, 'r', encoding='utf-8') as fin:
        reader = csv.reader(fin)
        for row in reader:
            if not row:
                continue
            cell = row[0].strip()
            # Skip header or invalid rows that don't start with a brace
            if not cell.startswith('{'):
                continue
            try:
                # Use ast.literal_eval to parse Python literal dictionaries
                data = ast.literal_eval(cell)
            except Exception as e:
                print("Error decoding data with ast.literal_eval:", e)
                continue

            # If the data contains an "entry" key that is a list, process each paper record.
            if "entry" in data and isinstance(data["entry"], list):
                for entry in data["entry"]:
                    out_row = process_entry(entry)
                    output_rows.append(out_row)

    with open(output_file, 'w', newline='', encoding='utf-8') as fout:
        # Using tab as the delimiter (change to ',' if desired)
        writer = csv.DictWriter(fout, fieldnames=headers, delimiter='\t')
        writer.writeheader()
        for row in output_rows:
            writer.writerow(row)

if __name__ == '__main__':
    input_csv = 'scopus_retracted_articles.csv'
    output_csv = 'output3.csv'

    process_csv(input_csv, output_csv)
    print("Processing complete. Output saved to", output_csv)
```

```python
import requests
import pandas as pd

# --- Existing Request Setup ---
API_KEY = "10a6fa24e6beeef124888369a691d5aa"
BASE_URL = "https://api.elsevier.com/content/search/scopus"

# List of DOIs from CSV
doi_csv = pd.read_csv(r"C:\Users\Parsa\OneDrive\Documents\Internship\Sadegh Suleimani Internship\Web Confrence\Files\retracted_compDataElec_DOI.csv")
dois = list(doi_csv["RetractionDOI"])

headers = {
    "X-ELS-APIKey": API_KEY,
    "Accept": "application/json"
}

# --- Function to Query Scopus API ---
def query_scopus_by_doi(doi):
    params = {
        "query": f"DOI({doi})",  # Query in the required format
        "view": "FULL"           # Request full record details
    }
    response = requests.get(BASE_URL, headers=headers, params=params)
    response.raise_for_status()  # Raise error if request failed
    return response.json()

# --- Request and Collect the JSON Responses ---
results = {}
for doi in dois:
    try:
        results[doi] = query_scopus_by_doi(doi)
    except Exception as e:
        print(f"Error for DOI {doi}: {e}")
        results[doi] = None

# --- Parse the Returned JSON Information ---
parsed_records = []
for doi, data in results.items():
    record = {"Queried_DOI": doi}  # Keep track of the DOI you queried
    # Check if data is valid and contains the expected keys
    if data and "search-results" in data and "entry" in data["search-results"] and data["search-results"]["entry"]:
        entry = data["search-results"]["entry"][0]  # Assume the first entry is correct

        # Extract some common fields; adjust field names as needed per the API docs.
        record["Scopus_DOI"]       = entry.get("prism:doi", None)
        record["Title"]            = entry.get("dc:title", None)
        record["Authors"]          = entry.get("dc:creator", None)
        record["PublicationName"]  = entry.get("prism:publicationName", None)
        record["CoverDate"]        = entry.get("prism:coverDate", None)
        record["AggregationType"]  = entry.get("prism:aggregationType", None)
        # You can add more fields here if required.
    else:
        record["Error"] = "No entry found or invalid response"
    parsed_records.append(record)

# --- Create a DataFrame from the Parsed Records ---
parsed_df = pd.DataFrame(parsed_records)
print(parsed_df.head())

# --- (Optional) Save the Parsed Data to a CSV File ---
parsed_df.to_csv("parsed_scopus_results.csv", index=False)
print("Parsed results saved to parsed_scopus_results.csv")
```
